**Task 7:** Write a more general unit conversion function that converts between all units of energy listed under the link below. The function should take arguments: for the original value, the original unit and the target unit. Implement the function in a way that the default target unit is "Wh". Verify that function by converting 200 'toe' to 'Wh'.

> https://www.iea.org/data-and-statistics/data-tools/unit-converter

> **Note:** You can also just pick three units to convert between if you don't feel like going through all combinations.

```python
PREFIXES = {"": 1, "k": 1e3, "M": 1e6, "G": 1e9, "T": 1e12, "P": 1e15}
PER_JOULE = {
    "J": 1,
    "cal": 0.2388,
    "Btu": 0.0009478,
    "Wh": 0.0002778,
    "toe": 2.388e-11,
    "tce": 3.412e-11,
}


def split_unit(unit):
    for base, per_joule in PER_JOULE.items():
        prefix = unit.removesuffix(base)
        if unit.endswith(base) and prefix in PREFIXES:
            return PREFIXES[prefix], per_joule
    raise ValueError(f"Unknown unit: {unit}")
```

```python
def convert_unit(value, from_unit, to_unit="Wh"):
    from_prefix, from_per_joule = split_unit(from_unit)
    to_prefix, to_per_joule = split_unit(to_unit)
    return value * from_prefix / from_per_joule * to_per_joule / to_prefix
```

```python
convert_unit(200, "toe")
```

```python
convert_unit(1, "ktoe", "GWh")
```
